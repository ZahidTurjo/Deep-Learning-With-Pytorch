```python
vocab={'<unk>':0}
```

```python
for token in Counter(tokens).keys():
  if token not in vocab:
    vocab[token]=len(vocab)
vocab
```

```python
input_sentences=document.split('\n')
print(input_sentences)
```

```python
def text_to_indices(sentence,vocab):
  numerical_sentence=[]
  for token in sentence:
    if token in vocab:
      numerical_sentence.append(vocab[token])
    else :
      numerical_sentence.append(vocab['<unk>'])
  return numerical_sentence
```

```python
input_numerical_sentences=[]
for sentence in input_sentences:
  input_numerical_sentences.append(text_to_indices(word_tokenize(sentence.lower()),vocab))
```

```python
X=padded_training_seq[:,:-1]
y=padded_training_seq[:,-1]
```

```python

class CustomDataset(Dataset):
  def __init__(self,X,y):
    self.X=X
    self.y=y

  def __len__(self):
    return self.X.shape[0]
  def __getitem__(self,index):
    return self.X[index],self.y[index]

```

```python
dataset=CustomDataset(X,y)
```

```python
len(dataset)
```

```python
datatloader=DataLoader(dataset,batch_size=32,shuffle=True)
```

```python
# create Model
class LSTMmodel(nn.Module):
  def __init__(self,vocab_szie):
    super().__init__()
    self.embedding=nn.Embedding(vocab_szie,100)
    self.lstm=nn.LSTM(100,150,batch_first=True)
    self.fc=nn.Linear(150,vocab_szie)
  def forward(self,x):
    embedded=self.embedding(x)
    inter_hs,(final_hs,final_cs)=self.lstm(embedded)
    output=self.fc(final_hs.squeeze(0))
    return output
```

```python
model=LSTMmodel(len(vocab))
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

```python
model.to(device)
```

```python
epochs = 50
learning_rate = 0.001

criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate)
```

```python
for epoch in range(epochs):
  total_loss=0
  for batch_x,batch_y in datatloader:
    batch_x,batch_y=batch_x.to(device), batch_y.to(device)

    optimizer.zero_grad()

    output=model(batch_x)

    loss=criterion(output,batch_y)

    loss.backward()

    optimizer.step()

    total_loss=total_loss+loss.item()

  print(f"Epoch : {epoch +1}--> loss : {total_loss}")
```

```python
def prediction(device,model, vocab, text):
    model.eval()

    tokenized_text = word_tokenize(text.lower())
    numerical_text = text_to_indices(tokenized_text, vocab)

    padded_text = torch.tensor(
        [0] * (max(len_list) - len(numerical_text)) + numerical_text,
        dtype=torch.long
    ).unsqueeze(0).to(device)   #  move to device

    with torch.no_grad():
        output = model(padded_text)

    value, index = torch.max(output, dim=1)

    predicted_word = list(vocab.keys())[index.item()]

    return text + " " + predicted_word


```

```python
prediction(device,model, vocab, "What is the Advanced Machine Learning")
```

```python
# Function to calculate accuracy
def calculate_accuracy(model, dataloader, device):
    model.eval()  # Set the model to evaluation mode
    correct = 0
    total = 0

    with torch.no_grad():  # No need to compute gradients
        for batch_x, batch_y in dataloader:
            batch_x, batch_y = batch_x.to(device), batch_y.to(device)

            # Get model predictions
            outputs = model(batch_x)

            # Get the predicted word indices
            _, predicted = torch.max(outputs, dim=1)

            # Compare with actual labels
            correct += (predicted == batch_y).sum().item()
            total += batch_y.size(0)

    accuracy = correct / total * 100
    return accuracy

# Compute accuracy
accuracy = calculate_accuracy(model, datatloader, device)
print(f"Model Accuracy: {accuracy:.2f}%")
```
